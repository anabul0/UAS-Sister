```python
from google.colab import drive
import rasterio
import matplotlib.pyplot as plt

# Hubungkan Google Drive
drive.mount('/content/drive')

# Tentukan path file GeoTIFF
tif_path = '/content/drive/My Drive/Klasifikasi_Landsat8_Jakarta.tif'

# Baca file GeoTIFF menggunakan Rasterio
with rasterio.open(tif_path) as src:
    classified_img = src.read(1)

# Visualisasikan hasil klasifikasi
plt.figure(figsize=(10, 6))
plt.imshow(classified_img, cmap='viridis')
plt.colorbar(label='Class ID')
plt.title("Hasil Klasifikasi Landsat 8")
plt.show()
```
